```python
import json

# Load existing data
try:
    with open("students.json", "r") as file:
        students = json.load(file)
except FileNotFoundError:
    students = {
        "ravi": 33,
        "ram": 55,
        "vansh": 46,
        "sohan": 67
    }

print("1. Press A - Add a student")
print("2. Press B - Update marks")
print("3. Press C - Search for a student")
print("4. Press D - Display all students and marks")

n = input("Enter your choice: ").lower()

# A -Add student
if n == "a":
    a = input("Enter the student name: ").lower()
    m = int(input("Enter the student marks: "))

    students[a] = m

    # Save permanently
    with open("students.json", "w") as file:
        json.dump(students, file, indent=4)

    print("Student added successfully!")


# B -Update marks
elif n == "b":
    k = input("Enter the student name: ").lower()

    if k in students:
        v = int(input("Enter the new marks: "))
        students[k] = v

        # Save permanently
        with open("students.json", "w") as file:
            json.dump(students, file, indent=4)

        print("Marks updated successfully!")
    else:
        print("Student not found!")


# C -Search student
elif n == "c":
    s = input("Enter the student name you want to search: ").lower()

    if s in students:
        print("Name :", s)
        print("Marks:", students[s])
    else:
        print("Student not found!")


# D -Display all students
elif n == "d":
    with open("students.json", "r") as file:
        students = json.load(file)

    print("Name     Marks")

    for name, marks in students.items():
        print(name, "  ", marks)

else:
    print("Wrong choice")
```
